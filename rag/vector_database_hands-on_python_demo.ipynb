##### Create a collection, add documents, and search by meaning.

```python
import chromadb

# 1. Create an in-memory client (data disappears when the script ends)
client = chromadb.Client()

# 2. Create a collection (like a table)
collection = client.create_collection(name="hello_chroma")
#collection = client.create_collection()
```

```python
# 3. Add documents. Chroma embeds them automatically.

# i) read file, ii) extract file content, iii) split text to chunk,

# HNSW ->

collection.add(
    documents=[
        "Pizza is an Italian dish with cheese and tomato.",
        "Python is a popular programming language.",
        "Dogs are loyal pets that love to play fetch.",
        "The Eiffel Tower is a famous landmark in Paris.",
    ],
    ids=["doc1", "doc2", "doc3", "doc4"],
)
```

```python
print(collection.get(ids=["doc1"]))
```

```python
print(collection.get(ids=["doc1"], include=["embeddings", "documents"]))
```

```python
data = collection.get()
print(data)
```

```python
data = collection.get(ids=["doc1", "doc2"], 
                      include=["embeddings", "documents"],)
print(data["documents"])
print(data["embeddings"])
print(len(data["documents"]))
print(len(data["embeddings"][0]))
print(len(data["embeddings"][1]))
```

```python
# 4. Search by meaning
results = collection.query(
    query_texts=["Which food comes from Italy?"],
    n_results=2,
)

print(results["documents"])
print(results["distances"])
```

```python
# 4. Search by meaning
results = collection.query(
    query_texts=["Which food comes from Italy?"],
    n_results=4,
)

print(results["documents"])
print(results["distances"])
```
