```md
This data is a token usage breakdown from a generative AI model's response metadata (specifically from Google's Gemini API). It tells you exactly how much computational "text" was processed to handle your request.
In AI, text is broken down into tokens, which are characters or chunks of characters (roughly 4 characters or 0.75 words per token).
Here is what each specific line means:
• candidates_token_count=59: This is the length of the final answer generated for you. The AI used 59 tokens to write its visible response.
• prompt_token_count=10: This is the length of your input prompt. Your request or question was very short, consisting of only about 10 tokens.
• thoughts_token_count=581: This represents the AI's internal reasoning or "chain of thought". Before giving the final answer, the model spent 581 tokens "thinking," planning, or reasoning through the problem behind the scenes.
• total_token_count=650: This is the grand total of all tokens used for this specific interaction (59 + 10 + 581 = 650). AI providers use this total number to calculate API costs or rate limits.
```

## 2. OpenAI's GPT LLM model

#### Connect the DeepSeek Model Family Using an OpenAI-Compatible API.

````python
%pip install openai
````

````python
# Import required packages

import os
from dotenv import load_dotenv
````

````python
load_dotenv()
````

````python
BASE_URL = os.getenv("OPENAI_BASE_URL", "https://api.apinex.bond/v1")
API_KEY = os.getenv("OPENAI_API_KEY", "sk-mytoken")
MODEL_NAME = os.getenv("MODEL_NAME", "free/deepseek-v4.1-flash")
````

````python
from openai import OpenAI

# The client reads OPENAI_API_KEY from your environment automatically
client = OpenAI(api_key=API_KEY, base_url=BASE_URL)

# 1. User prompt
user_prompt = "Explain generative AI in 3 simple sentences."
````

````python
# 2. Send the prompt to the LLM
# base_url="https://api.openai.com/v1"
response = client.responses.create(
    model=MODEL_NAME,
    input=user_prompt
)
````

````python
response
````

````python
# 3 & 4. Get the generated content and show it to the user
print(response.output_text)
````

# Streaming version

````python
stream = client.chat.completions.create(
    model=MODEL_NAME,
    messages=[{"role": "user", "content": "Write a short poem about the sea."}],
    stream=True,
)

for chunk in stream:
    if chunk.choices and chunk.choices[0].delta.content:
        print(chunk.choices[0].delta.content, end="", flush=True)
````

#### Connect the OpenAI's GPT(Open-weight model) Model Family Using an OpenAI-Compatible API.

````python
# Import required packages
import os
from openai import OpenAI

HF_BASE_URL = os.getenv("HF_BASE_URL", "https://router.huggingface.co/v1")
HF_TOKEN = os.getenv("HF_TOKEN", "hf_mytoken")
HF_MODEL_NAME = os.getenv("HF_MODEL_NAME", "openai/gpt-oss-20b:groq")
````

````python
# The client reads HF_TOKEN from your environment automatically
client = OpenAI(
    base_url=HF_BASE_URL,
    api_key=os.environ["HF_TOKEN"],
)
````

````python
# 2. Send the prompt to the LLM
completion = client.chat.completions.create(
    model=HF_MODEL_NAME,
    messages=[
        {
            "role": "user",
            "content": "What is the capital of France?"
        }
    ],
)
````

````python
completion
````

````python
# 3 & 4. Get the generated content and show it to the user
print(completion.choices[0].message)
````

````python
# Import required packages
import os
from openrouter import OpenRouter
````

````python
# Initialize the client (It automatically picks up OPENROUTER_API_KEY from env if not passed)
with OpenRouter(
    api_key=os.getenv("OPENROUTER_API_KEY"),
) as client:

    # Send a chat completion request
    response = client.chat.send(
        model="google/gemma-4-26b-a4b-it:free", # Example model ID
        messages=[
            {"role": "user", "content": "Hello! How do I use OpenRouter in Python?"}
        ]
    )

    print(response.choices[0].message.content)
````

# Using OpenAI compatible API

````python

import os
from openai import OpenAI

client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=os.environ["OPENROUTER_API_KEY"],  # get a key at openrouter.ai/keys
)
````

````python
response = client.chat.completions.create(
    model="google/gemma-4-26b-a4b-it:free",  # any model slug from openrouter.ai/models
    messages=[
        {"role": "user", "content": "Explain what an API is in one sentence."}
    ],
)

print(response.choices[0].message.content)
````
